```python
import os

for root, dirs, files in os.walk("."):
    print(root)
    for f in files:
        if f.endswith(".csv"):
            print("   ", f)
```

```python
import pandas as pd

customers = pd.read_csv("../real_estate_pipeline/data/raw/customers.csv")
accounts = pd.read_csv("../real_estate_pipeline/data/raw/accounts.csv")
transactions = pd.read_csv("../real_estate_pipeline/data/raw/transactions.csv")
loans = pd.read_csv("../real_estate_pipeline/data/raw/loans.csv")
```

```python
datasets = {
    "Customers": customers,
    "Accounts": accounts,
    "Transactions": transactions,
    "Loans": loans
}

for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("=" * 50)
    print(f"Rows : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")
    print(df.head())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.info())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.duplicated().sum())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.describe(include="all"))
```
